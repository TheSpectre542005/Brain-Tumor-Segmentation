# 04 — Attention U-Net 3D Training (Proposed Model)
**BraTS 2023 GLI Brain Tumor Segmentation | RCOEM 2026–27**

This notebook trains the **proposed Attention-Augmented 3D U-Net** with:
- **Squeeze-and-Excitation blocks** (channel attention) at every encoder/decoder level
- **Spatial Attention Gates** at all 4 decoder stages

Everything else mirrors the baseline training for fair comparison.

**Estimated training time on 1 DGX GPU:** ~12–14 hours

```python
import sys, os

# ─────────────────────────────────────────────────────────────
DATASET_PATH  = '/home/agrawalrb_1/Brain_Tumor_Segmentation/ASNR-MICCAI-BraTS2023-GLI-Challenge-TrainingData'  # ← CHANGE THIS
QUICK_TEST    = False
RESUME        = True
FOLD          = 0

# Ablation: enable/disable attention types
USE_CHANNEL_ATTN = True   # Squeeze-and-Excitation blocks
USE_SPATIAL_ATTN = True   # Spatial attention gates
# ─────────────────────────────────────────────────────────────

PROJECT_ROOT = os.path.abspath('..')
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from src.config import (
    BATCH_SIZE, NUM_EPOCHS, LEARNING_RATE, WEIGHT_DECAY,
    MAX_PATIENTS, N_FOLDS, DEVICE, RANDOM_SEED,
    EARLY_STOP_PATIENCE, LR_PATIENCE, LR_FACTOR
)
print(f'Device   : {DEVICE}')
print(f'Channel Attention (SE) : {USE_CHANNEL_ATTN}')
print(f'Spatial Attention Gates: {USE_SPATIAL_ATTN}')
```

```python
import time
import numpy as np
import torch
import matplotlib.pyplot as plt
from torch.optim import Adam

from src.models  import AttentionUNet3D
from src.losses  import CombinedLoss
from src.dataset import get_dataloaders
from src.utils   import (set_seed, save_checkpoint, load_checkpoint,
                         train_one_epoch, validate_one_epoch,
                         get_scheduler, get_writer, format_time)

set_seed(RANDOM_SEED)
print('✅ All imports OK')
```

## 1️⃣ Load Data

```python
train_loader, val_loader, test_loader, test_folders = get_dataloaders(
    dataset_path = DATASET_PATH,
    batch_size   = BATCH_SIZE,
    num_workers  = 4,
    fold         = FOLD,
    n_folds      = N_FOLDS,
    max_patients = MAX_PATIENTS,
    quick_test   = QUICK_TEST,
    quick_test_n = 10,
)
print(f'Train batches: {len(train_loader)}')
print(f'Val batches  : {len(val_loader)}')
```

## 2️⃣ Build Attention U-Net Model

```python
# Auto-name the model based on which attention types are enabled
if USE_CHANNEL_ATTN and USE_SPATIAL_ATTN:
    MODEL_NAME = 'attention_unet3d_full'
elif USE_CHANNEL_ATTN:
    MODEL_NAME = 'attention_unet3d_channel_only'
elif USE_SPATIAL_ATTN:
    MODEL_NAME = 'attention_unet3d_spatial_only'
else:
    MODEL_NAME = 'unet3d_no_attention'

model = AttentionUNet3D(
    in_channels      = 4,
    out_channels     = 4,
    init_features    = 32,
    use_channel_attn = USE_CHANNEL_ATTN,
    use_spatial_attn = USE_SPATIAL_ATTN,
).to(DEVICE)

total_params = sum(p.numel() for p in model.parameters())
print(f'Model      : {MODEL_NAME}')
print(f'Parameters : {total_params:,}')

optimizer = Adam(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
criterion = CombinedLoss(num_classes=4, dice_weight=0.5, focal_weight=0.5)
scheduler = get_scheduler(optimizer, patience=LR_PATIENCE, factor=LR_FACTOR)
writer    = get_writer(MODEL_NAME)
print('✅ Model ready')
```

## 3️⃣ Resume From Checkpoint (if available)

```python
start_epoch = 0
best_val_dice = 0.0

if RESUME:
    start_epoch, best_val_dice = load_checkpoint(
        model, optimizer, model_name=MODEL_NAME, prefer_best=False
    )

max_epochs = 3 if QUICK_TEST else NUM_EPOCHS
print(f'Training from epoch {start_epoch} → {max_epochs}')
```

## 4️⃣ Training Loop

```python
train_losses, val_losses, val_dices = [], [], []
no_improve_count = 0
training_start   = time.time()

for epoch in range(start_epoch, max_epochs):
    epoch_start = time.time()

    train_metrics = train_one_epoch(
        model, train_loader, optimizer, criterion, DEVICE, epoch, writer
    )
    val_metrics = validate_one_epoch(
        model, val_loader, criterion, DEVICE, epoch, writer
    )

    scheduler.step(val_metrics['wt_dice'])

    is_best = val_metrics['wt_dice'] > best_val_dice
    if is_best:
        best_val_dice = val_metrics['wt_dice']
        no_improve_count = 0
    else:
        no_improve_count += 1

    save_checkpoint(model, optimizer, epoch, val_metrics['wt_dice'],
                    model_name=MODEL_NAME, is_best=is_best)

    train_losses.append(train_metrics['loss'])
    val_losses.append(val_metrics['loss'])
    val_dices.append(val_metrics['wt_dice'])

    epoch_time = time.time() - epoch_start
    eta = epoch_time * (max_epochs - epoch - 1)
    print(f'Epoch [{epoch+1:3d}/{max_epochs}] '
          f'| Train: {train_metrics["loss"]:.4f} '
          f'| Val: {val_metrics["loss"]:.4f} '
          f'| WT Dice: {val_metrics["wt_dice"]:.4f} '
          f'| Best: {best_val_dice:.4f} '
          f'| ETA: {format_time(eta)}')

    if no_improve_count >= EARLY_STOP_PATIENCE:
        print(f'⚠️  Early stopping at epoch {epoch+1}')
        break

writer.close()
print(f'\n🏁 Training complete! Best Val WT Dice: {best_val_dice:.4f}')
print(f'   Total time: {format_time(time.time() - training_start)}')
```

## 5️⃣ Plot Training Curves

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
epochs_ran = range(1, len(train_losses) + 1)

axes[0].plot(epochs_ran, train_losses, label='Train Loss', color='#e74c3c', lw=2)
axes[0].plot(epochs_ran, val_losses,   label='Val Loss',   color='#3498db', lw=2)
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Loss')
axes[0].set_title(f'Loss — {MODEL_NAME}')
axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].plot(epochs_ran, val_dices, label='Val WT Dice', color='#9b59b6', lw=2)
axes[1].axhline(y=max(val_dices), color='gray', ls='--', alpha=0.7,
                label=f'Best: {max(val_dices):.4f}')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Dice Score')
axes[1].set_title(f'Val WT Dice — {MODEL_NAME}')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f'training_curves_{MODEL_NAME}.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Saved: training_curves_{MODEL_NAME}.png')
print('\n🏁 Notebook 04 complete. Proceed to 05_evaluate_compare.ipynb')
```
